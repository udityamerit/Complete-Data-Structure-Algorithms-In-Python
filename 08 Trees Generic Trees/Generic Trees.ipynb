## Printing the Tree Nodes

```python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.children = []

root = TreeNode(1)

child1 = TreeNode(2)
child2 = TreeNode(3)
child3 = TreeNode(4)
child4 = TreeNode(5)

root.children.append(child1)
root.children.append(child2)
root.children.append(child3)
root.children.append(child4)

def print_tree(root):
    print(root.data)
    for childNode in root.children:
        print_tree(childNode)

print_tree(root)
```

## Printing the tree in the better way

```python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.children = []

root = TreeNode(1)

child1 = TreeNode(2)
child2 = TreeNode(3)
child3 = TreeNode(4)
child4 = TreeNode(5)
child5 = TreeNode(6)


root.children.append(child1)
root.children.append(child2)
root.children.append(child3)
root.children.append(child4)
root.children.append(child5)

def print_tree_better(root):
    if root is None:
        return

    print(root.data, end=":")

    for eachChild in root.children:
        print(eachChild.data, end=",")

    print()

    for eachChild in root.children:
        print_tree_better(eachChild)


print_tree_better(root)

child2.children.append(child3)
child2.children.append(child5)

child4.children.append(child3)
child4.children.append(child2)

print_tree_better(root)
```

## Take the Input recursivilly 

```python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.children = []

def print_tree_better(root):
    if root is None:
        return

    print(root.data, end=":")

    for eachChild in root.children:
        print(eachChild.data, end=",")

    print()

    for eachChild in root.children:
        print_tree_better(eachChild)


def take_input():
    data  = int(input("Enter the data for the node: "))

    Node = TreeNode(data)

    num_child = int(input(f"Enter the no of child for the {data}: "))

    for _ in range(num_child):
        child = take_input()
        Node.children.append(child)

    return Node

root = take_input()

print_tree_better(root)
```

## Take input Level Wise

```python
from collections import deque

class TreeNode:
    def __init__(self, data):
        self.data = data
        self.children = []

def print_tree_better(root):
    if root is None:
        return
    print(root.data, end=":")
    for eachChild in root.children:
        print(eachChild.data, end=",")
    print()
    for eachChild in root.children:
        print_tree_better(eachChild)

def take_input_level_wise():
    data = int(input("Enter the root Node data: "))
    root = TreeNode(data)

    queue = deque([root])

    while queue:
        current_node = queue.popleft()

        num_children = int(input(f"Enter the number of Children for {current_node.data}: "))

        for i in range(num_children):
            child_data = int(input(f"Enter the data for child {i+1} of Node {current_node.data}: "))
            child_node = TreeNode(child_data)
            current_node.children.append(child_node)
            queue.append(child_node)

    return root


root = take_input_level_wise()
print("\n--- Final Tree Structure ---")
print_tree_better(root)
```
