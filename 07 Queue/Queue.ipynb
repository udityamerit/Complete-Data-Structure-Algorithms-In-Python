```python
class Queue:
    def __init__(self):
        self.__queue=[]

    def size(self):
        return f"len of the ques is: {len(self.__queue)}"

    def isEmpty(self):
        return self.size()==0

    def enqueue(self, data):
        self.__queue.append(data)
        return f"data is added in the Queue {data}"

    def front(self):
        if self.size()==0:
            print("Queue is empty, can't show front")
            return None
        return f"front element is: {self.__queue[0]}"

    def dequeue(self):
        if self.size()==0:
            print("Queue is empty, can't dequeue")
            return None
        return f"deques the element: {self.__queue.pop(0)}"


Q = Queue()

Q.enqueue(10)
Q.enqueue(20)
Q.enqueue(30)
Q.enqueue(40)

print(Q.size())
print(Q.isEmpty())
print(Q.front())
print(Q.dequeue())
print(Q.dequeue())
print(Q.front())

# print(Q.self.__queue)
```
