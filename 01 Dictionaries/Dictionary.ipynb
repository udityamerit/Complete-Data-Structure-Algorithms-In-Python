```python
# Nested dicts

students={
    "student1":{'name':'uditya', 'age':23},
    "student2":{'name':'peter', 'age':43}
}
print(students)
```

```python
print(students['student1']['name'])
print(students['student2']['age'])
```

```python
# iterating over the nested dict

for std_id, std_info in students.items():
    print(f"{std_id}:{std_info}")
    for key, value in std_info.items():
        
        print(key, value)
```

# Merge two sorted lists

```python
def merge_two_sorted_lists(list1, list2):
    ans_list = []
    i = 0
    j = 0
    
    # Traverse both lists and add the smaller element
    while i < len(list1) and j < len(list2):
        if list1[i] <= list2[j]:
            ans_list.append(list1[i])
            i += 1
        else:
            ans_list.append(list2[j])
            j += 1
            
    # Append any remaining elements from list1
    while i < len(list1):
        ans_list.append(list1[i])
        i += 1
        
    # Append any remaining elements from list2
    while j < len(list2):
        ans_list.append(list2[j])
        j += 1
        
    return ans_list
```

# merge three dictionaries

```python
def merge_three_dictionaries(dict1, dict2, dict3):
    for key, value in dict2.items():
        dict1[key] = value

    for key, value in dict3.items():
        dict1[key] = value

    return dict1
```

# merge_dicts_with_overlapping_keys

```python
def merge_dicts_with_overlapping_keys(dicts):
    # Your code goes here
    d = {}
    for i in range(len(dicts)):
        for k, v in dicts[i].items():
            if k not in d:
                d[k]=v
            else:
                d[k]+=v 
    return d 
dicts = [{'a': 1, 'b': 2}, {'b': 3, 'c': 4}, {'c': 5, 'd': 6}]
print(merge_dicts_with_overlapping_keys(dicts))
```

# is subset

```python
def is_subset(lst1, lst2):
    for i in range(len(lst1)):
        found = False

        for j in range(len(lst2)):
            if lst1[i] == lst2[j]:
                found = True
                break

        if not found:
            return False

    return True
```
