```python
class Stack:
    def __init__(self):
        self.stack = []

    def push(self, data):
        self.stack.append(data)
        print(f"{data} pushed to stack")


    def size(self):
        return f"size of the stack is: {len(self.stack)}"

    def is_empty(self):
        return len(self.stack) == 0

    def top(self):
        if not self.is_empty():
            return f"Top element of the stack is: {self.stack[-1]}"
        else:
            return None

    def pop(self):
        if not self.is_empty():
            return f"poping the element from the stack: {self.stack.pop()}"
        else:
            return None


mystack = Stack()
print(mystack.is_empty())      
mystack.push(1)
mystack.push(2)
mystack.push(3)
mystack.push(4)
print(mystack.is_empty())       
print(mystack.pop())            
print(mystack.pop())            
print(mystack.size())           
print(mystack.top())  

print(mystack.stack.insert(0,10))
print(mystack.stack)
```
