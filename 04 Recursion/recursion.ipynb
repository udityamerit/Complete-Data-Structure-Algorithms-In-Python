# Recursion

```python
## finding 2 ki power n using recursion

def power2(n):
    if n==1:
        return 2
    smallAns  = power2(n-1)
    ans = 2*smallAns

    return ans

print(power2(5))
print("hello")
```

```python
def sumN(n):
    if n==0:
        return 0

    ans = n+sumN(n-1)
    # print(ans)
    return ans

print(sumN(5))
```

## Remove the char using recursion

```python
def remove_char(s, ch):
    if len(s) == 0 or s == "":
        return s
    smallans = remove_char(s[1:], ch)
    if s[0] == ch:
        return smallans
    else:
        return s[0] + smallans

s = "abczzz"
ans = remove_char(s, 'z')
print(ans)  
```

## Checking the string is polindrom or not

```python
def helper_function(s, start, end):
    if(start>=end):
        return True

    if(s[start]!=s[end]):
        return False

    return helper_function(s, start+1, end-1)


def polindrom(s):
    return helper_function(s, 0, len(s)-1)


print(polindrom("abcdcba"))
print(polindrom("abcdgae"))



def is_palindrome(s):
    if len(s) <= 1:
        return True
    if s[0] != s[-1]:
        return False
    return is_palindrome(s[1:-1])   
```

## printing the subsequences

```python
def subsequence(s):
    if s=="":
        ans = [""]
        return ans

    smallans = subsequence(s[1:])
    mychar = s[0]
    ans = []
    ans.extend(smallans)
    for each in smallans:
        ans.append(mychar+each)

    return ans

s= "abc"
print(subsequence(s))
```
