# Recursion

```python
## finding 2 ki power n using recursion

def power2(n):
    if n==1:
        return 2
    smallAns  = power2(n-1)
    ans = 2*smallAns

    return ans

print(power2(5))
```

```python
def sumN(n):
    if n==0:
        return 0

    ans = n+sumN(n-1)
    # print(ans)
    return ans

print(sumN(5))
```
