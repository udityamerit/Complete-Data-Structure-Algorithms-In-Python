```python
## positional arguments for the variable lengths
def print_numbers(*args):
    for number in args:
        print(number)

print_numbers(1,2,3,4,5)
```

```python
## positional arguments for the variable lengths
def print_numbers(**kwargs):
    for key, value in kwargs.items():
        print(key, value)

print_numbers(name="uditya", age='3')
```

```python
lst = [1,2,3,4,5]
k =6
def reverse(lst):
    if len(lst)%k==0:
        return lst
    else:
        for i in range(k):
            c = lst[-1]
            lst.insert(0,c)
            # print(lst)
            del lst[-1]
            # print(lst)
    print(lst)

reverse(lst)
```
