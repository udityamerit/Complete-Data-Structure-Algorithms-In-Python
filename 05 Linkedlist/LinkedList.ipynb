# LinkedList

## Taking Input For LinkedList

```python
class Node:
    def __init__(self, value):
        self.data = value 
        self.next = None


## taking input

def taking_input():
    value = int(input("Enter the value: "))
    head = None

    while(value!=-1):
        newnode = Node(value)
        if head==None:
            head = newnode
        else:
            temp = head
            while temp.next is not None:
                temp = temp.next
            temp.next = newnode

        value = int(input("Enter the value: "))
    return head


## printing the LinkedList

def print_LL(head):
    temp = head
    while temp is not None:
        print(temp.data, end="->")
        temp = temp.next

    return

newhead=taking_input()
print_LL(newhead)
```

# Inserting the node at position

```python
class Node:
    def __init__(self, value):
        self.data = value
        self.next = None


def take_input():
    val = int(input("Enter the data: "))
    head = None
    tail = None

    while val!=-1:
        newNode = Node(val)

        if head is None:
            head = newNode
            tail = newNode
        else:
            tail.next = newNode
            tail = newNode

        val = int(input("Enter the data: "))

    return head


def insert_at_head(head, data):
    newNode = Node(data)
    newNode.next = head
    head = newNode

    return head


def insert_at_tail(head, data):
    newNode = Node(data)
    temp = head

    if head is None:
        return newNode

    while temp.next is not None:
        temp = temp.next

    temp.next = newNode

    return head


def insert_at_position(head, data, index):
    if index==0:
        return insert_at_head(head, data)

    newNode = Node(data)
    temp = head
    count = 0

    while temp is not None and count<index-1:
        temp = temp.next
        count+=1

    if temp is None:
        print("Index Out of Range Check Your Index")

    newNode.next = temp.next
    temp.next = newNode

    return head


def print_LL(head):
    temp = head
    while temp is not None:
        print(temp.data, end="-->")
        temp = temp.next

    print("None")


value = take_input()
print("Linked List:")
print_LL(value)

print("Inserted at head: ")
he = insert_at_head(value, 40)
print_LL(he)

print("Inserted at tail: ")
ta = insert_at_tail(he, 400)
print_LL(ta)

print("Inserted at an index of 3: ")
bet = insert_at_position(ta, 300, 3)
print_LL(bet)
```
