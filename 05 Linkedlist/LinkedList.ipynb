## Taking Input For LinkedList

```python
class Node:
    def __init__(self, value):
        self.data = value 
        self.next = None


## taking input

def taking_input():
    value = int(input("Enter the value: "))
    head = None

    while(value!=-1):
        newnode = Node(value)
        if head==None:
            head = newnode
        else:
            temp = head
            while temp.next is not None:
                temp = temp.next
            temp.next = newnode

        value = int(input("Enter the value: "))
    return head


## printing the LinkedList

def print_LL(head):
    temp = head
    while temp is not None:
        print(temp.data, end="->")
        temp = temp.next

    return

newhead=taking_input()
print_LL(newhead)
```

#### Cycle Detection in LL

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

def input_ll():
    val = int(input("Enter the data: "))
    head = None
    tail = None
    while val!=-1:
        node = Node(val)
        if head is None:
            head = node

            tail = node
        else:
            tail.next = node
            tail = node

        val = int(input("Enter the data: "))

    return head

def print_ll(head):
    temp = head
    while temp is not None:
        print(temp.data, end="-->")
        temp =temp.next

    print("None")

def has_cycle(head):
    if head is None or head.next is None:
        return f"LinkedList does not contains cycle "
        
    slow = head
    fast = head.next
    
    while fast is not None and fast.next is not None:
        if fast == slow:
            return f"LinkedList contains cycle "
            
        fast = fast.next.next
        slow = slow.next
            
    return f"LinkedList does not contains cycle "
```

### Checking the LinkedList is polindrom or not

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def isPalindrome(head: ListNode) -> bool:
    if not head or not head.next:
        return True
        
    # Step 1: Find the middle of the linked list
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        
    # Step 2: Reverse the second half of the list
    prev = None
    curr = slow
    while curr:
        next_node = curr.next
        curr.next = prev
        prev = curr
        curr = next_node
        
    # Step 3: Compare both halves
    left, right = head, prev
    while right:  # The right half will be shorter or equal in length
        if left.val != right.val:
            return False
        left = left.next
        right = right.next
        
    return True
```

#### Removing all the elements

```python
# Definition for singly linked list node.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def remove_elements(head, val):
    """
    Function to remove all nodes with value val from the linked list.
    :param head: ListNode -> head of the singly linked list
    :param val: int -> the value to be removed
    :return: ListNode -> the head of the new linked list
    """
    # 1. Create a dummy node pointing to the head
    dummy = ListNode(0)
    dummy.next = head
    
    # 2. Initialize the traversal pointer at the dummy node
    current = dummy
    
    # 3. Traverse the list, checking the NEXT node's value
    while current.next is not None:
        if current.next.val == val:
            # Skip the next node to remove it from the list
            current.next = current.next.next
        else:
            # Move to the next node only if we didn't delete anything
            current = current.next
            
    # 4. Return the actual head of the modified list
    return dummy.next

```
